## Quiz 4 進階討論：不同條件下的拼接效果

### 1. 幾乎沒有重疊區的兩張圖片是無法拼接成功的
僅以配對數作為成功判準並不足夠，重疊 40% 案例雖通過門檻，但 RANSAC 內點比例僅 0.03，顯示配對品質完全不可信，凸顯內點比例才是更關鍵的可信度指標

### 2. 角度問題對於拼接效果沒什麼影響
實測了旋轉20、40、60、80、100度，RANSAC內點比例依舊差不多，實際的拼接效果也很好

### 3. 亮度過暗的照片找不到特徵點，會導致拼接失敗
以亮度變暗90%為例，只有找到一個特徵點，代表亮度過暗到一種程度，會找不到特徵點
